#**1. Data Cleaning**

```python
# Create a copy of the original DataFrame to avoid modifying it directly
df = df.copy()

# Remove the first row based on its integer position using .iloc
# This is useful when the first row contains metadata or is not part of the actual data
df = df.iloc[1:].copy()

# Display the structure of the DataFrame after removing the first row
df.info()

# Display the first five rows of the updated DataFrame to verify the changes
df.head()
```

```python
# Remove duplicate rows from the DataFrame to ensure data integrity
df = df.drop_duplicates()

# Display concise summary of the DataFrame including column types and non-null counts
df.info()
```

```python
df_date = df[['year', 'active_year', 'date_prec', 'source_date', 'date_start', 'date_end']]
df_date.info()
df_date.head()
```

```python
# Convert the 'year' column to datetime format, specifying the year format explicitly
df_date['year'] = pd.to_datetime(df_date['year'].astype(str), format='%Y')

# Extract the year as an integer from the datetime object and overwrite the 'year' column
df_date['year'] = df_date['year'].dt.year

# Display concise summary of the DataFrame including data types and non-null counts
df_date.info()

# Show the first few rows of the DataFrame to verify the transformation
df_date.head()
```

```python
# Assuming your series is called `dates`
df_date['date_start'] = pd.to_datetime(df_date['date_start'])  # Convert to datetime if not already

# Extract year, month, and day
df_date['year_date_start'] = df_date['date_start'].dt.year
df_date['month_date_start'] = df_date['date_start'].dt.month
df_date['day_date_start'] = df_date['date_start'].dt.day

# Optional: Display result
df_date[['date_start', 'year_date_start', 'month_date_start', 'day_date_start']].head()
```

```python
# Proper way to drop 'date_start' column
df_date.drop('date_start', axis=1, inplace=True)
```

```python

# Assuming your series is called `dates`
df_date['date_end'] = pd.to_datetime(df_date['date_end'])  # Convert to datetime if not already

# Extract year, month, and day
df_date['year_date_end'] = df_date['date_end'].dt.year
df_date['month_date_end'] = df_date['date_end'].dt.month
df_date['day_date_end'] = df_date['date_end'].dt.day

# Optional: Display result
df_date[['date_end', 'year_date_end', 'month_date_end', 'day_date_end']].head()
```
